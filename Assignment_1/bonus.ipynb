Dictionary of four lists of different datatypes.

```python
data_dict = {
    "discrete": [2, 5, 10, 60, 4, 22, 18, 98, 55, 32, 1, 7],
    "continuous": [0.5, 5.54, 2.4, 7.213, 9.9, 15.67, 3.14],
    "nominal": ["TypeScript", "Java", "TypeScript", "Python", "Kotlin", "Python", "C++", "Java", "TypeScript"],
    "ordinal": ["January", "February", "February", "March", "March", "March", "April", "May", "May", "June"]
    }
```

Function that returns the most frequently ocurring item in a list using pythons max function.

```python
def categorical_mode(dict, column_name):
    cat_list = dict[column_name]

    return max(cat_list)
```

Getting the mode of two lists.

```python
nominal_mode = categorical_mode(data_dict, "nominal")
print(nominal_mode)
```

```python
ordinal_mode = categorical_mode(data_dict, "ordinal")
print(ordinal_mode)
```
