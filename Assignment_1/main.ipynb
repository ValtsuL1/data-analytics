Dictionary of four lists of different datatypes.

```python
data_dict = {
    "discrete": [2, 5, 10, 60, 4, 22, 18, 98, 55, 32, 1, 7],
    "continuous": [0.5, 5.54, 2.4, 7.213, 9.9, 15.67, 3.14],
    "nominal": ["TypeScript", "Java", "TypeScript", "Python", "Kotlin", "Python", "C++", "Java", "TypeScript"],
    "ordinal": ["January", "February", "February", "March", "March", "March", "April", "May", "May", "June"]
    }
```

Function that returns the mean of a numerical list by summing its values and dividing by its length.

```python
def numerical_mean(dict, column_name):
    num_list = dict[column_name]

    return sum(num_list) / len(num_list)
```

Getting the mean of two lists of differing datatypes.

```python
discrete_mean = numerical_mean(data_dict, "discrete")

print(discrete_mean)
```

```python
continuous_mean = numerical_mean(data_dict, "continuous")

print(continuous_mean)
```
