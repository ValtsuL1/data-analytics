```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib import colors
from matplotlib.ticker import PercentFormatter
import seaborn as sns

df = sns.load_dataset('iris')

df.loc[df['species'] == 'setosa', 'species_id'] = 0
df.loc[df['species'] == 'versicolor', 'species_id'] = 1
df.loc[df['species'] == 'virginica', 'species_id'] = 2

df
```

Pairplot with matplotlib

```python
n_bins = 20

fig, axs = plt.subplots(4, 4, figsize=(10,10), layout='constrained')
axs[0, 0].hist(df['sepal_length'], bins=n_bins)
axs[0, 0].set_ylabel('sepal_length')
axs[0, 1].scatter(df['sepal_width'], df['sepal_length'], c=df['species_id'], cmap='Set3')
axs[0, 2].scatter(df['petal_length'], df['sepal_length'], c=df['species_id'], cmap='Set3')
axs[0, 3].scatter(df['petal_width'], df['sepal_length'], c=df['species_id'], cmap='Set3')

axs[1, 0].scatter(df['sepal_length'], df['sepal_width'], c=df['species_id'], cmap='Set3')
axs[1, 0].set_ylabel('sepal_width')
axs[1, 1].hist(df['sepal_width'], bins=n_bins)
axs[1, 2].scatter(df['petal_length'], df['sepal_width'], c=df['species_id'], cmap='Set3')
axs[1, 3].scatter(df['petal_width'], df['sepal_width'], c=df['species_id'], cmap='Set3')

axs[2, 0].scatter(df['sepal_length'], df['petal_length'], c=df['species_id'], cmap='Set3')
axs[2, 0].set_ylabel('petal_length')
axs[2, 1].scatter(df['sepal_width'], df['petal_length'], c=df['species_id'], cmap='Set3')
axs[2, 2].hist(df['petal_length'], bins=n_bins)
axs[2, 3].scatter(df['petal_width'], df['petal_length'], c=df['species_id'], cmap='Set3')

axs[3, 0].scatter(df['sepal_length'], df['petal_width'], c=df['species_id'], cmap='Set3')
axs[3, 0].set_ylabel('petal_width')
axs[3, 0].set_xlabel('sepal_length')
axs[3, 1].scatter(df['sepal_width'], df['petal_width'], c=df['species_id'], cmap='Set3')
axs[3, 1].set_xlabel('sepal_width')
axs[3, 2].scatter(df['petal_length'], df['petal_width'], c=df['species_id'], cmap='Set3')
axs[3, 2].set_xlabel('petal_length')
axs[3, 3].hist(df['petal_width'], bins=n_bins)
axs[3, 3].set_xlabel('petal_width')



plt.show()
```

Pairplot with seaborn

```python
graph = sns.pairplot(df, hue='species',vars=['sepal_length', 'sepal_width', 'petal_length', 'petal_width'])

plt.show()
```
