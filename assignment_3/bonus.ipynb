International overnight stays in Lapland in 2024

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('nights_spent_international_2024.csv')

fig, axs = plt.subplots()
axs.pie(df['Nights_spent'], labels=df['Country'], startangle=45, textprops={'size': 'smaller'})
```
