1.3 Random vector of length 15, and its mean and standard deviation.

```python
while True:
    a = np.random.randint(1, 15)
    b = np.random.randint(1, 15)
    c = np.random.randint(1, 15)

    length = np.sqrt(a*a + b*b + c*c)
    if length == 15:
        vector = np.array([a, b, c])
        
        print(vector)
        print(f"Mean deviation: {np.mean(vector)}")
        print(f"Standard deviation: {np.std(vector)}")
        
        break
    
```

2.1 Survival rate by gender and travel class

```python
import seaborn as sns

titanic = sns.load_dataset('titanic')
titanic
```

Survival rate by gender

```python
survival_by_gender = titanic.groupby('sex')['survived'].mean() * 100
survival_by_gender
```

Survival rate by travel class

```python
survival_by_class = titanic.groupby('class')['survived'].mean() * 100
survival_by_class
```

2.2 Average fare for survivors

```python
survival_by_fare = titanic.groupby(['class', 'survived'])['fare'].mean()
survival_by_fare
```
