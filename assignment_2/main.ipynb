1.1 100x100 matrix, mean, sum, min and max

```python
import numpy as np
```

```python
matrix = np.random.randint(0, 10000, (100, 100))
print(matrix)
```

```python
mean = matrix.mean()
print(mean)
```

```python
sum = matrix.sum()
print(sum)
```

```python
max = matrix.max()
print(max)
```

```python
min = matrix.min()
print(min)
```

1.2 10000 dice rolls

```python
rolls_list = []

for i in range(10000):
    rolls_list.append(np.random.randint(1, 7))

rolls = np.array(rolls_list)

for i in range(1, 7):
    print(f"{i} was rolled {np.bincount(rolls)[i]} times.")
```
