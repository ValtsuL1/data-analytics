1.2 10000 dice rolls

```python
rolls_list = []

for i in range(10000):
    rolls_list.append(np.random.randint(1, 7))

rolls = np.array(rolls_list)

for i in range(1, 7):
    print(f"{i} was rolled {np.bincount(rolls)[i]} times.")
```

Assignment 2

```python
import pandas as pd
```

```python
data = pd.read_csv("koeti_riista_tau_001_en.csv", encoding='iso-8859-1')
data
```

2.1 Grouping by month

```python
accidents_by_month = data.value_counts('month', sort=False)
accidents_by_month
```

2.2 Grouping by roadtype and species

```python
accidents_by_roadtype = data.value_counts(['roadTypeShort', 'wildlifeSpeciesName'], sort=False)
accidents_by_roadtype
```

2.3 Accidents by day of week and hour, and combined

```python
data["weekDay"] = pd.to_datetime(data["eventTime"], utc=True).dt.day_of_week
data["hour"] = pd.to_datetime(data["eventTime"], utc=True).dt.hour
```

```python
accidents_by_weekday = data.value_counts('weekDay', sort=False)
accidents_by_weekday
```

```python
accidents_by_hour = data.value_counts('hour', sort=False)
accidents_by_hour
```

```python
accidents_by_hour_of_weekday = data.value_counts(['weekDay', 'hour'], sort=False)
pd.options.display.max_rows = 200
accidents_by_hour_of_weekday
```
